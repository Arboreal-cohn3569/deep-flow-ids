```python
{
 "cells": [
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "# 02. Comparative Benchmarks & Metrics Visualizer\n",
    "Evaluate inference latencies and accuracy metrics across saved models."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "import time\n",
    "import numpy as np\n",
    "import pandas as pd\n",
    "import tensorflow as tf\n",
    "\n",
    "X_test = np.load('../data/processed/X_test.npy')\n",
    "y_test = np.load('../data/processed/y_test.npy')\n",
    "\n",
    "model_names = ['1D_CNN', 'RNN_LSTM', 'Hybrid_CNN_LSTM']\n",
    "benchmark_results = []\n",
    "\n",
    "for name in model_names:\n",
    "    model_path = f'../assets/{name}.keras'\n",
    "    model = tf.keras.models.load_model(model_path)\n",
    "    \n",
    "    start_time = time.time()\n",
    "    preds = model.predict(X_test, batch_size=512)\n",
    "    latency = (time.time() - start_time) / len(X_test) * 1000  # ms/sample\n",
    "    \n",
    "    acc = np.mean(np.argmax(preds, axis=1) == y_test)\n",
    "    benchmark_results.append({'Model': name, 'Accuracy': acc, 'Latency (ms/sample)': latency})\n",
    "\n",
    "df_res = pd.DataFrame(benchmark_results)\n",
    "display(df_res)"
   ]
  }
 ],
 "metadata": {
  "language_info": {
   "name": "python"
  }
 },
 "nbformat": 4,
 "nbformat_minor": 2
}
```
